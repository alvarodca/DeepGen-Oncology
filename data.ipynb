# Dataset Extraction

100.064 instances on h5ad files on CELLxGENE data which can be found here: https://cellxgene.cziscience.com/collections/65db5560-7aeb-4c66-b150-5bd914480eb8


```python
import urllib.request

# URl of the actual dataset, takes a few mins to load
url = "https://datasets.cellxgene.cziscience.com/36ab5d3d-158e-4988-9700-e14fc7102fe4.h5ad"
urllib.request.urlretrieve(url, "wu_2021_breast_cancer.h5ad")
```

```python
import scanpy as sc # pip install scanpy, used for h5ad files

data = sc.read_h5ad("wu_2021_breast_cancer.h5ad")
```
